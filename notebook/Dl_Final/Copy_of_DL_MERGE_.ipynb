```python
import pandas as pd
df1=pd.read_csv('/content/drive/MyDrive/D-BOOKS - 1 DL (1).csv')
df2=pd.read_csv('/content/drive/MyDrive/D-BOOKS - 2 DL.csv')
df3=pd.read_csv('/content/drive/MyDrive/D-BOOKS - 3 DL.csv')
df4=pd.read_csv('/content/drive/MyDrive/D-BOOKS - 4 DL.csv')
df5=pd.read_csv('/content/drive/MyDrive/D-BOOKS - 5 DL.csv')
df6=pd.read_csv('/content/drive/MyDrive/D-BOOKS - 6 DL.csv')
df7=pd.read_csv('/content/drive/MyDrive/D-BOOKS - 7 DL.csv')
```

```python
print('df1',df1.columns)
print('df2',df2.columns)
print('df3',df3.columns)
print('df4',df4.columns)
print('df5',df5.columns)
print('df6',df6.columns)
print('df7',df7.columns)
```

SWAPPING COLUMNS

```python
cols=list(df4.columns)
i,j=cols.index('rating'),cols.index('review')
cols[i],cols[j]=cols[j],cols[i]
df4=df4[cols]
```

```python
cols=list(df5.columns)
i,j=cols.index('rating'),cols.index('review')
cols[i],cols[j]=cols[j],cols[i]
df5=df5[cols]
```

```python
cols=list(df6.columns)
i,j=cols.index('rating'),cols.index('review')
cols[i],cols[j]=cols[j],cols[i]
df6=df6[cols]
```

```python
cols=list(df7.columns)
i,j=cols.index('rating'),cols.index('review')
cols[i],cols[j]=cols[j],cols[i]
df7=df7[cols]
```

MERGING

```python
DL_MERGE_df=pd.concat([df1,df2,df3,df4,df5,df6,df7],ignore_index=True)
```
