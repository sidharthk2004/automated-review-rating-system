*Word Clouds for High vs. Low Ratings*

```python
from wordcloud import WordCloud

high_reviews = df[df['review/score'] >= 4]['review/text']
low_reviews = df[df['review/score'] <= 2]['review/text']

WordCloud(width=800, height=400).generate(" ".join(high_reviews.dropna()))
WordCloud(width=800, height=400).generate(" ".join(low_reviews.dropna()))
```

Sentiment vs. Rating Consistency

```python
from textblob import TextBlob

df['sentiment'] = df['review/text'].apply(lambda x: TextBlob(str(x)).sentiment.polarity)
sns.scatterplot(data=df, x='sentiment', y='review/score', alpha=0.3)
plt.title("🧠 Sentiment vs. Rating")
plt.show()
```

```python
Review Length vs. Rating
```

```python
df['length'] = df['review/text'].apply(lambda x: len(str(x).split()))
sns.boxplot(data=df, x='review/score', y='length')
plt.title("📝 Review Length vs. Rating")
plt.show()
```

Helpfulness vs. Rating

```python
sns.boxplot(data=df, x='review/score', y='review/helpfulness')
plt.title("👍 Helpfulness vs. Rating")
plt.yscale('log')  # Optional for skewed values
plt.show()
```

Category-Level Rating Trends

```python
category_avg = df.groupby('categories')['review/score'].mean().sort_values()
category_avg.plot(kind='barh', figsize=(10, 50), title="📚 Avg Rating by Category")
plt.xlabel("Average Score")
plt.show()
```

```python
df.drop_duplicates(subset=['review','rating'],inplace=True)
```

```python
print(df.duplicated().sum())
```

```python


import re

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"[^\w\s]", "", text)
    text = re.sub(r"\d+", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df.loc[:, "review"] = df["review"].apply(clean_text)
```

```python
df['rating']=df['rating'].astype(int)
```
